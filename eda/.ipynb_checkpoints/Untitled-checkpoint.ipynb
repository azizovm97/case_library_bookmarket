```python
# Заголовки и описание
cells = [
    nbf.v4.new_markdown_cell("# 📊 Exploratory Data Analysis (EDA)\n\nАнализ данных библиотеки."),
    nbf.v4.new_code_cell("import pandas as pd\nimport duckdb\nimport plotly.express as px"),
    nbf.v4.new_code_cell("# Подключение к базе данных\ncon = duckdb.connect('../my.db')\n\nbooks = con.execute('SELECT * FROM books_list').df()\ngenres = con.execute('SELECT * FROM genre').df()\nreviews = con.execute('SELECT * FROM book_reviews').df()"),
    nbf.v4.new_code_cell("# Объединение таблиц\nbooks_with_genre = books.merge(genres, on='genre_id')"),
    nbf.v4.new_markdown_cell("## 📈 Диаграммы"),

    # График 1: книги по жанрам
    nbf.v4.new_code_cell("fig_books_by_genre = px.histogram(books_with_genre, x='genre_name', title='Количество книг по жанрам')\nfig_books_by_genre.show()"),

    # График 2: книги по годам
    nbf.v4.new_code_cell("fig_books_by_year = px.histogram(books_with_genre, x='year_published', title='Распределение книг по годам публикации')\nfig_books_by_year.show()"),

    # График 3: топ авторов
    nbf.v4.new_code_cell("fig_top_authors = px.bar(books_with_genre['author'].value_counts().head(10), title='Топ-10 авторов по количеству книг')\nfig_top_authors.show()"),

    # График 4: топ по рейтингу
    nbf.v4.new_code_cell(\"\"\"ratings_df = con.execute('''
SELECT b.title, AVG(r.rating) AS avg_rating
FROM book_reviews r
JOIN books_list b ON r.book_id = b.book_id
GROUP BY b.title
ORDER BY avg_rating DESC
LIMIT 10
''').df()\n
fig_top_rated = px.bar(ratings_df, x='avg_rating', y='title', orientation='h', title='Топ-10 книг по рейтингу')\nfig_top_rated.show()\"\"\")
]
```
